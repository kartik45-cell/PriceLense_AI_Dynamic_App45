```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
if str(ROOT) not in sys.path: sys.path.insert(0, str(ROOT))
from src.data_processing import load_raw_data, clean_transactions
raw, meals = load_raw_data(ROOT / 'data')
df = clean_transactions(raw, meals)
print(df.shape, df.week.min(), df.week.max())

from src.feature_engineering import make_model_frame
frame = make_model_frame(df)
cutoff = frame.week.max() - 12
print({'train_rows': int((frame.week <= cutoff).sum()), 'test_rows': int((frame.week > cutoff).sum()), 'cutoff_week': int(cutoff)})
```
